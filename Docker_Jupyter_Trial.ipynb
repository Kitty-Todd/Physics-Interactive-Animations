```python
import ipywidgets as widgets
import json
import os
from IPython.display import display, Video

lab_choice = widgets.Dropdown(options=[("Projectile Motion", "ProjectileSim"), ("Friction", "Friction")], value="ProjectileSim", description="Select Lab: ")

g_slider = widgets.FloatSlider(value=9.8, min=1.0, max = 20.0, description="Gravity: ")

v_slider = widgets.FloatSlider(value=5.0, min=1.0, max=15.0, description="Velocity: ")
proj_vars = widgets.VBox([v_slider])

m_slider = widgets.FloatSlider(value=1.0, min=0.5, max=5.0, description="Mass: ")
s_slider = widgets.FloatSlider(value=0.5, min=0.0, max=1.0, description="Static Friction Coeff: ")
k_slider = widgets.FloatSlider(value=0.3, min=0.0, max=1.0, description="Kinetic Friction Coeff (please keep lower than static): ")
F_app_slider = widgets.FloatSlider(value=10.0, min=0.0, max=110.0, description="Max Applied Force: ")
fric_vars = widgets.VBox([m_slider, s_slider, k_slider, F_app_slider])

var_container = widgets.Box([proj_vars])

def on_lab_change(change):
    if change["new"] == "proj":
        var_container.children = [proj_vars]
    else:
        var_container.children = [fric_vars]

lab_choice.observe(on_lab_change, names="value")

run_btn = widgets.Button(description="Launch Simulation", button_style="primary")
output = widgets.Output()

def run_sim(b):
    with output:
        output.clear_output()

        data = {"lab": lab_choice.value, "gravity": g_slider.value, "velocity": v_slider.value, "mass": m_slider.value, "static_friction": s_slider.value, "kinetic_friction": k_slider.value, "max_force": F_app_slider.value}
        
        with open("params.json", "w") as f:
            json.dump(data, f)
        
        print("Rendering you custom physics scene...")

        scene_name = "ProjectileSim" 
        if lab_choice.value == "proj" :
            scene_name = "ProjectileSim"
        else:
            scene_name = "Friction"

        os.system(f"manim -ql -v WARNING phys_sim.py {scene_name}")

        video_path = f"media/videos/phys_sim/480p15/{scene_name}.mp4"
        if os.path.exists(video_path):
            display(Video(video_path, embed=True, width=600))
        else:
            print("Error: Video not found. Check your script names!")

run_btn.on_click(run_sim)

display(widgets.HTML("<h2>Interactive Physics Platform<h2>"))
display(lab_choice)
display(g_slider)
display(var_container)
display(run_btn, output)
```
